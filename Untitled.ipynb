```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python

df = pd.read_csv('arquivos/gas_prices.csv')


plt.title('Valor de combustível por ano (galão) em USD')
plt.plot(df['Year'], df['Australia'], 'r.-', label= 'Australia')
plt.plot(df['Year'], df['Italy'], 'y.-', label='Italia')
plt.plot(df['Year'], df['USA'], 'b.-', label='USA')
plt.plot(df['Year'], df['Japan'], 'g.-', label='Japao')
plt.ylabel('Preco')
plt.xlabel('Ano')
plt.xticks(df['Year'][::2])

plt.legend()
plt.show()
```
