```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python

df = pd.read_csv('arquivos/gas_prices.csv')


plt.title('Valor de combustível por ano (galão) em USD')
paises = ['Australia', 'USA', 'Canada', 'Mexico', 'Brazil', 'France', 'Germany']
for p in df:
    if p in paises:
        plt.plot(df['Year'], df[p], label = p, marker='.')
plt.ylabel('Preco')
plt.xlabel('Ano')
plt.xticks(df['Year'][::2])

plt.legend()
plt.savefig('grafico_exemplo.jpeg')
plt.show()
```
